```python
# inspect the unit column
df['unit'].unique()
```

```python
import re

def parse_unit(unit):

    if pd.isna(unit):
        return 'unknown', np.nan

    unit = str(unit).strip().lower()

    # Handle count-based units
    if unit in ['unit', 'head', 'bunch']:
        return unit, 1.0

    # Handle weight and volume units
    m = re.fullmatch(
        r'(\d+(?:\.\d+)?)?\s*(kg|g|l|ml)',
        unit
    )

    if not m:
        return 'other', np.nan

    qty = float(m.group(1)) if m.group(1) else 1.0

    base, factor = {
        'kg': ('kg', 1),
        'g': ('kg', 0.001),
        'l': ('l', 1),
        'ml': ('l', 0.001)
    }[m.group(2)]

    return base, qty * factor
```

```python
# apply the parse_unit function to the 'unit' column and create new columns for base unit and base quantity
parsed = df['unit'].map(parse_unit)

df['base_unit'] = parsed.map(lambda x: x[0])
df['base_qty'] = parsed.map(lambda x: x[1])

df['price_per_base'] = df['price'] / df['base_qty']
```

```python
# validate the results of the parsing and calculations
print("Missing base_qty:", df['base_qty'].isna().sum())

print("\nBase unit distribution:")
print(df['base_unit'].value_counts())

print("\nMissing price_per_base:", df['price_per_base'].isna().sum())
```
